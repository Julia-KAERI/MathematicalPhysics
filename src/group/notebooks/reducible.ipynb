```julia
DI = [1 0; 0 1]
D123=[ω 0; 0 ωp]
D132=[ωp 0; 0 ω]
D12 = [0 1;1 0]
D13 = [0 ω;ωp 0]
D23 = [0 ωp; ω 0]
```

```julia
Ds = [DI, D123, D132, D12, D13, D23]

sum([conj((X)[2, 1]) * X[2, 1] for X in Ds])
```
