```julia
using Symbolics
```

```julia
@variables x y z θx θy θz
```

```julia
Rx = [1 0 0.0; 0 cos(θx) sin(θx); 0 -sin(θx) cos(θx)]
Ry = [-sin(θy) 0 cos(θy); 0 1 0; cos(θy) 0 sin(θy)]
Rz = [cos(θz) sin(θz) 0; -sin(θz) cos(θz) 0; 0 0 1];
```

```julia
Rx*Ry
```

```julia
Ry*Rx
```

```julia
Rx*Ry - Ry*Rx
```

```julia
R = [sin(θy) 0 cos(θy); 0 1 0; cos(θy) 0 -sin(θy)]
S = [cos(θz) -sin(θz) 0; sin(θz) cos(θz) 0; 0 0 1];
```

```julia
R*S*[x ; y ; z]
```

```julia
Q=R*S*[x;y;z]
expand.(Q[1]*Q[2])
```
